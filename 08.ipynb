```python
import numpy as np #handles array
import matplotlib.pyplot as plt #draw clusteering graph
from sklearn.cluster import AgglomerativeClustering #Imports the Agglomerative Clustering algorithm from Scikit-learn.

# Simple dataset
X = np.array([
    [1, 2],
    [2, 3],
    [2, 1],
    [8, 8],
    [9, 8],
    [8, 9]
])

# Hierarchical Clustering
model = AgglomerativeClustering(n_clusters=2)
labels = model.fit_predict(X) #perform clustering

# Print cluster labels
print("Cluster Labels:", labels)

# Plot
plt.scatter(X[:,0], X[:,1], c=labels, s=100)
plt.xlabel("X")
plt.ylabel("Y")
plt.title("Hierarchical Clustering")
plt.show()
```
