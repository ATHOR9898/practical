```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.cluster import KMeans

# Real-life dataset: Student Performance
# [Study Hours, Exam Marks]

X = np.array([
    [1, 35],
    [2, 40],
    [2, 45],
    [3, 50],
    [3, 55],
    [4, 60],
    [4, 65],
    [5, 70],
    [5, 75],
    [6, 80],
    [6, 85],
    [7, 90]
])

# Number of clusters
k = 3

# Create K-Means model
kmeans = KMeans(n_clusters=k, random_state=42, n_init=10)

# Fit the model
kmeans.fit(X)

# Get cluster labels
labels = kmeans.labels_

# Get centroids
centroids = kmeans.cluster_centers_

# Display results
print("Cluster Labels:")
print(labels)

print("\nCluster Centroids:")
print(centroids)

# Display each student and cluster
print("\nStudent and Cluster:")

for i in range(len(X)):
    print(
        "Student:", i + 1,
        "Study Hours:", X[i][0],
        "Marks:", X[i][1],
        "-> Cluster:", labels[i] + 1
    )

# Plot clusters
plt.scatter(X[:, 0], X[:, 1], c=labels, s=100)

# Plot centroids
plt.scatter(
    centroids[:, 0],
    centroids[:, 1],
    marker='X',
    s=200
)

plt.xlabel("Study Hours")
plt.ylabel("Exam Marks")
plt.title("K-Means Clustering - Student Performance")
plt.show()
```
