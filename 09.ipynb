```python
# Install required library
# pip install mlxtend

import pandas as pd #handle the dataset.
from mlxtend.preprocessing import TransactionEncoder #converts transactions into binary form.
from mlxtend.frequent_patterns import apriori, association_rules #finds frequent itemsets,generates rules.

# Step 1: Create transaction dataset
transactions = [
    ['Milk', 'Bread', 'Butter'],
    ['Bread', 'Butter'],
    ['Milk', 'Bread'],
    ['Milk', 'Bread', 'Butter', 'Eggs'],
    ['Bread', 'Eggs'],
    ['Milk', 'Butter'],
    ['Bread', 'Butter', 'Eggs'],
    ['Milk', 'Bread', 'Butter']
]

# Step 2: Convert transactions into one-hot encoded format. The transactions are converted into a table containing True or False.
te = TransactionEncoder()
te_array = te.fit(transactions).transform(transactions)

df = pd.DataFrame(te_array, columns=te.columns_)

print("Transaction Dataset:")
print(df)

# Step 3: Generate frequent itemsets
frequent_itemsets = apriori(
    df,
    min_support=0.3, #means an itemset must occur in at least 30% of transactions.
    use_colnames=True
)

print("\nFrequent Itemsets:")
print(frequent_itemsets)

# Step 4: Generate association rules Only rules having confidence ≥ 60% are generated.
rules = association_rules(
    frequent_itemsets,
    metric="confidence",
    min_threshold=0.6
)

# Step 5: Display important measures
rules = rules[
    ['antecedents', 'consequents', 'support',
     'confidence', 'lift']
]

print("\nAssociation Rules:")
print(rules)
```
