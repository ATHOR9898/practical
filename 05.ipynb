```python
# Data Discretization using Equal Width Binning

import pandas as pd

# Sample Engineering Student Data
data = {
    'Student': ['Amit', 'Priya', 'Rahul', 'Sneha', 'Rohan',
                'Neha', 'Kiran', 'Pooja', 'Akash', 'Anjali'],
    'Marks': [35, 48, 56, 62, 67, 73, 81, 88, 91, 96]
}

# Create DataFrame
df = pd.DataFrame(data)

print("Original Data")
print(df)

# Divide marks into 3 equal-width bins
df['Performance'] = pd.cut(df['Marks'],
                           bins=3,
                           labels=['Low', 'Medium', 'High'])

print("\nData after Discretization")
print(df)
```
